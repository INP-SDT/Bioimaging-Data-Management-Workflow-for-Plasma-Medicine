# Bioimaging Data Management Workflow for Plasma Medicine

In automation-assisted fluorescence imaging, experiments are usually organized in multi-well plates (e.g., 96-well or 384-well). Each well contains a specific condition (e.g., different drug, concentration, treatment conditions, or replicate). A plate groups these wells together in a defined layout, and a screen consists of multiple plates that make up the entire experimental campaign.
<div style="margin-left: 0em; display: table;">
    <ul>
        <li><span>Screen level = overall plasma treatment study (e.g., “Plasma exposure parameter screen on adherent cells”).</span></li><br>
        <li><span>Plate level = one physical culture plate defined by a specific plasma parameter set (e.g., “Plate 2: nozzle–cell distance series (5–20 mm) at fixed treatment time”).</span></li><br>
        <li><span>Well level = one experimental unit corresponding to a single plasma condition (e.g., “Well C6: A375 cells, nozzle distance 10 mm, plasma exposure 40 s, replicate 1”).</span></li>
    </ul>
</div>
This workflow provides a structured path to ensure that bioimaging experiments are not only carried out but also documented and managed in a FAIR-compliant way. To achieve this, our workflow guides you through the process of metadata collection, annotation, and analysis for bioimaging experiments. 
It integrates three key components: 
<div style="margin-left: 0em; display: table;">
    <ul>
    <li><strong>Adamant (metadata schema):</strong></li>
        Adamant provides structured, machine-readable forms (based on community standards such as Plasma-MDS or REMBI) to capture experimental metadata. Here, you define details about the screen, plates, and wells, including treatments, conditions, and replicates. This ensures consistency and semantic interoperability.
    <li><strong>eLabFTW (electronic lab notebook):</strong></li>
eLabFTW is the human-readable record of your experiment. It connects experimental protocols, raw data references, and the metadata generated in Adamant. By linking directly to OMERO datasets, it acts as the central hub for documenting the why and how of your experiment, complementing the what captured in metadata.
    <li><strong>OMERO (image data management and annotation):</strong></li>
OMERO stores and organizes the raw image data. By combining uploaded images with Adamant-generated metadata, OMERO annotations make images searchable, comparable, and interpretable. At the screen, plate, and well levels, annotation ensures that each image is linked to the biological and experimental context in which it was generated.
    </ul>
</div>

## Step-by-Step Workflow

**1. Carry out your bioimaging experiment** (e.g., automation-assisted fluorescence imaging, drug screening, live-cell imaging, etc.)

<div style="border: 2px solid black; padding: 6px; border-radius: 6px; margin-left: 6em; display: table;">
    <strong> Examples </strong> <br>
    Best-suited experiment types for screen–plate–well model:
    <ul>
        <li><strong>Automation-assisted fluorescence imaging :</strong>
            <ul>
            <li>Cell imaging in multi-well plates</li>
            <li>Different wells correspond to treatments, drugs, etc. </li>
            </ul>
        </li>
        <li><strong>Drug screening / Dose–response studies:</strong>
            <ul>
            <li>Multiple compounds tested across wells at varying concentrations</li>
            <li>Plates organized by drug panels; the screen captures the full library</li>
            </ul>
        </li>
        <li><strong>Live-cell imaging in multi-well formats:</strong>
            <ul>
            <li>Time-lapse imaging of different conditions side-by-side in one plate</li>
            <li>Each well is a separate time series but linked under the same plate</li>
            </ul>
        </li>
    </ul>
    <strong style="color:black">Note</strong>
    <ul style="color:black;">
        <li>Keep track of essential variables:</li>
        <ul>
            <li>Experimental design (controls, treatments, replicates)</li>
            <li>Biological material (cell type, tissue, organism)</li>
            <li>Imaging conditions (microscope settings, time points, magnification)</li>
        </ul>
    </ul>
    <span style="color:black"> Store raw results in a secure local directory for later upload </span> <br>
</li>
</div>
<br>

**2. Fill the biological metadata** (REMBI Template)
FAIR data reuse require structured, standardized descriptors. REMBI (Recommended Metadata for Biological Images) ensures interoperability across labs and repositories. <br>
2.1 Download the [REMBI-compatible Excel template](https://github.com/INP-SDT/Bioimage-Data-Workflow-for-Plasma-Medicine/blob/main/metadataschemas/biologicalmetadata.xlsx) provided by your lab or consortium
<div style="border: 2px solid black; padding: 6px; border-radius: 6px; margin-left: 6em; display: table;">
    <strong style="color:black"> Note </strong> <br>
    <span style="color:black"> Enter information such as: </span> <br>
    <ul style="color:black;">
        <li><span> Biological system: species, strain, cell line, tissue </span></li>
        <li><span> Sample preparation: media, reagents, treatment concentrations </span></li>
        <li><span> Microscopy details: objective, exposure, channels </span></li>
    </ul>    
</div>
<br>

2.2 Save the file as .xlsx and keep it with your raw data

**3. Enter experiment metadata in Adamant**
<div style="border: 2px solid black; padding: 6px; border-radius: 6px; margin-left: 6em; display: table;">
    <strong> Explanation </strong> <br>
    <span> Adamant translates your experimental details into structured, machine-readable metadata (JSON). In this step, we make sure that data are consistent and semantically aligned with community standards (e.g., Plasma-MDS, REMBI).</span>
</div>
    
3.1 Open [Adamant](https://plasma-mds.github.io/adamant/) <br>
3.2 Click on "BROWSE SCHEMA" <br>
3.3 Load the [screenschema.json](https://github.com/INP-SDT/Bioimage-Data-Workflow-for-Plasma-Medicine/blob/main/metadataschemas/screenschema.json) into Adamant <br>
3.4 Click on "RENDER" <br>
3.5 Fill in all metadata fields with your experimental details <br>
3.6 Click on “PROCEED” (bottom-right corner of the form) to send the metadata to eLabFTW <br>
<div style="border: 2px solid black; padding: 6px; border-radius: 8px; margin-left: 6em; display: table;">
    <strong style="color:red;">Caution</strong><br>
    <span style="color:red"> An eLabFTW API key is required. Please log in to your ELN before using Adamant integration.</span>
</div>
<br>

**4. Document the Experiment in eLabFTW**
<div style="border: 2px solid black; padding: 6px; border-radius: 6px; margin-left: 6em; display: table;">
    <strong> Explanation </strong> <br>
    <span> eLabFTW acts as your human-readable record of what happened in the lab. This step links protocols, notes, and Adamant metadata so others (and your future self) can reproduce the experiment.</span>
</div>
<br>

4.1 Log into [eLabFTW](https://demo.elabftw.net/login.php) <br>
4.2 Create a new experiment entry with the same project identifier used in Adamant <br>
4.3 Attach the files <br>
<div style="border: 2px solid black; padding: 6px; border-radius: 6px; margin-left: 6em; display: table;">
    <strong style="color:black"> Note </strong> <br>
    <span style="color:black"> These files include:</span> <br>
    <ul style="color:black;">
        <li><span> Protocols and raw data references </span></li>
        <li><span> The REMBI Excel file </span></li>
        <li><span> Metadata JSON exported from Adamant (this can bew done automatically)</span></li>
    </ul>
</div>
<br>

4.4 Cross-link to your future OMERO dataset/project ID

**5. Upload images and Screen creation in OMERO** 
<div style="border: 2px solid black; padding: 6px; border-radius: 6px; margin-left: 6em; display: table;">
   <strong> Explanation </strong> <br>
   <span> OMERO is the central hub for managing image data. Uploading images with linked metadata ensures they are not just pictures but scientifically interpretable datasets.</span> 
</div>
<br>

5.1 Log into your OMERO instance [OMERO](https://demo.openmicroscopy.org/webclient/login/?url=%2Fwebclient%2F) to create a Screen for the annotation <br>
5.2 Create a new "Screen" <br>
5.3 Start the software OMERO.insight ([download available here](https://www.openmicroscopy.org/omero/downloads/)) <br>
5.4 Connect OMERO.insight to your OMERO instance <br>
5.5 Upload the taken images (from your experiment) into the in step 5.2 created Screen via OMERO.insight
<br>

6.5  Provide information needed for the annotation of the screen and plate and wells (if wanted). 
<div style="border: 2px solid black; padding: 6px; border-radius: 6px; margin-left: 6em; display: table;">
    <strong> Explanation </strong> <br>
    <span> In the annotation step, OMERO requires structured information about the screen, plates, and wells so that experimental conditions can be linked directly to the corresponding image data. This information typically includes screen identifiers, plate layouts, and well-specific treatments or replicates. To streamline this process, a code snippet is provided that only needs to be executed once. Running it creates an input field where users can enter the relevant details. After this initial setup, you no longer need to regenerate the input form—simply fill in the field with the required metadata and then run the subsequent code blocks, which will correctly interpret and attach your input to the respective screens, plates, and wells in OMERO. This approach reduces redundancy, ensures consistency in annotation, and makes the process both efficient and reproducible.  </span> 
</div>
<br>

<div style="border: 2px solid black; padding: 6px; border-radius: 8px; margin-left: 6em; display: table;">
    <strong style="color:red;">Caution</strong><br>
    <span style="color:red"> Run the following line of code only once to generate the entry fields.</span>
</div>

```python
# display fields for annotation information
screen = widgets.Text(description="Name of the screen that shall be annotated:", layout=widgets.Layout(width='600px'), style={'description_width': 'initial'})
plate = widgets.Text(description="Name of the plate that shall be annotated:", layout=widgets.Layout(width='600px'), style={'description_width': 'initial'})
experimentMetadata = widgets.Text(description="ID of the eLab experiment containing the metadata:", layout=widgets.Layout(width='600px'), style={'description_width': 'initial'})
checkbox = widgets.Checkbox(
    value=False,
    description='Annotate wells?'
)
output = widgets.Output()
# define output on checkbox click
def on_checkbox_change(change):
    with output:
        clear_output()  # vorherige Ausgabe löschen
        if change['new']:
            print("Wells will be annotated")
        else:
            print("Wells will not be annotated")

checkbox.observe(on_checkbox_change, names='value')
# display all widgets
display(screen)
display(plate)
display(experimentMetadata)
display(checkbox, output)
```

6.6 Run the following annotation code to finally annotate the extracted metadata to the provided screen, plate and well.

```python
# annotate metadata from eLab to Omero plate and screen
spinner = widgets.HTML('<i class="fa fa-spinner fa-spin fa-2x fa-fw"></i> Code processing...')
display(spinner)
# set data access to group "users" 
connection.setGroupForSession(3)
metadata = ELN.extract_experiment_attachments(expID = int(experimentMetadata.value), filetype = "JSON metadata")
bioData = ELN.extract_experiment_attachments(expID = int(experimentMetadata.value), filetype = "Excel file")

OME.annotate_screen(conn = connection, screenName = screen.value, jsonData = metadata)
OME.annotate_plate(conn = connection, plateName = plate.value, jsonData = metadata, excelData = bioData, wellBool = checkbox.value)

spinner.value = '<span style="color:green; font-weight:bold;">✅ Done!</span>'
```
